```python
import pandas as pd
import numpy as np 

df=pd.read_csv("StudentsPerformance.csv")

print(df.head())
```

```python
print(df.columns)
print(df.shape)
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
#average 
print("The average of math score:",df["math score"].mean())
print("The average of reading score:",df["reading score"].mean())
print("The average of writing score:",df["writing score"].mean())

#maximum
print("The maximum of math score:",df["math score"].max())
print("The maximum of reading score:",df["reading score"].max())
print("The maximum of writing score:",df["writing score"].max())

#minimum
print("The minimum of math score:",df["math score"].min())
print("The minimum of reading score:",df["reading score"].min())
print("The minimum of writing score:",df["writing score"].min())
```

```python
df["Average"]=((df["math score"]+
    df["reading score"]+
    df["writing score"])/3).round(2)

print(df.head())
```

```python
highest=df["Average"].max()

print(df[df["Average"]==highest])
```

```python
print(df.groupby("gender")["Average"].mean().round(2))
```

```python
df.to_csv("Average_studentPerformance.csv",index=False)
```
